````python
# Import standard library modules
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

# Import third-party modules
import numpy as np
import pandas as pd
import pandas_ta as ta
import matplotlib.pyplot as plt

# Import Alpaca modules
import alpaca
from alpaca.data.historical.crypto import CryptoHistoricalDataClient
from alpaca.data.requests import (
    CryptoBarsRequest,
    CryptoLatestQuoteRequest,
    CryptoQuoteRequest,
    CryptoTradesRequest,
)
from alpaca.data.timeframe import TimeFrame, TimeFrameUnit
from alpaca.trading.client import TradingClient
from alpaca.trading.enums import (
    AssetClass,
    AssetStatus,
    OrderSide,
    OrderType,
    QueryOrderStatus,
    TimeInForce,
)
from alpaca.trading.requests import LimitOrderRequest, MarketOrderRequest
````

````python
import os
API_KEY = os.getenv("ALPACA_API_KEY", "<YOUR PAPER API KEY>")
API_SECRET = os.getenv("ALPACA_API_SECRET", "<YOUR PAPER API SECRET>")
PAPER = os.getenv("ALPACA_PAPER", "True").lower() in ("true", "1", "yes")
trade_api_url = os.getenv("ALPACA_TRADE_API_URL", None)
trade_client = TradingClient(api_key=API_KEY, secret_key=API_SECRET, paper=PAPER, url_override=trade_api_url)
crypto_historical_data_client = CryptoHistoricalDataClient(api_key=API_KEY, secret_key=API_SECRET)
````

# Step 2: Executing Market Data Analysis for BTC/USD

````python
symbol = "BTC/USD"
timezone = ZoneInfo("America/New_York")
today = datetime.now(timezone).date()
start_date = today - timedelta(days=365 * 2)
end_date = today - timedelta(days=1)
req = CryptoBarsRequest(
    symbol_or_symbols = [symbol],
    timeframe=TimeFrame.Day,
    start=start_date,
    end=end_date,
)
bars_data = crypto_historical_data_client.get_crypto_bars(req).df.reset_index(level='symbol', drop=True)
````

## Calculating EMA & Indicators

````python
df = bars_data.copy()
short_window = 12
long_window = 26
df["ema_short"] = df["close"].ewm(span=short_window, adjust=False).mean()
df["ema_long"] = df["close"].ewm(span=long_window, adjust=False).mean()
sma_window = 50
df["sma"] = df["close"].rolling(window=sma_window).mean()
df.ta.atr(length=14, append=True)
df.ta.adx(length=14, append=True)
````

# Step 3: Generating Swing Trading Signals & Orders

````python
currently_above = df["ema_short"] > df["ema_long"]
previously_below_or_equal = df["ema_short"].shift(1) <= df["ema_long"].shift(1)
df["buy_signal"] = currently_above & previously_below_or_equal
currently_below = df["ema_short"] < df["ema_long"]
previously_above_or_equal = df["ema_short"].shift(1) >= df["ema_long"].shift(1)
df["sell_signal"] = currently_below & previously_above_or_equal
````

# Step 4: Integrating with Gemini AI & Sovereign OS API for Smart Execution

````python
import os
import json
import requests
import google.generativeai as genai
from alpaca.trading.requests import MarketOrderRequest
from alpaca.trading.enums import OrderSide, TimeInForce

# Configure Gemini AI
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY", "<YOUR GEMINI API KEY>")
genai.configure(api_key=GEMINI_API_KEY)

# Get latest market data and indicators
latest_row = df.iloc[-1]
latest_price = latest_row["close"]
ema_short_val = latest_row["ema_short"]
ema_long_val = latest_row["ema_long"]
adx_val = latest_row["ADX_14"] if "ADX_14" in latest_row else "N/A"
atr_val = latest_row["ATRr_14"] if "ATRr_14" in latest_row else "N/A"

# Prepare prompt for Gemini AI
prompt = f"""
Analyze the following technical indicators for BTC/USD and determine the optimal swing trading action.
Current Price: {latest_price}
EMA 12 (Short): {ema_short_val}
EMA 26 (Long): {ema_long_val}
ADX (14): {adx_val}
ATR (14): {atr_val}

Respond ONLY with a valid JSON object containing the following keys:
- "signal": "BUY", "SELL", or "HOLD"
- "confidence": a number between 0 and 100
- "reasoning": a brief explanation of the decision
"""

try:
    model = genai.GenerativeModel('gemini-2.5-flash')
    response = model.generate_content(prompt)
    res_text = response.text.strip()
    if res_text.startswith("```json"):
        res_text = res_text[7:]
    if res_text.endswith("```"):
        res_text = res_text[:-3]
    res_text = res_text.strip()
    decision = json.loads(res_text)
    print("Gemini AI Decision:", json.dumps(decision, indent=2))
    
    # Execute order based on Gemini AI decision
    signal = decision.get("signal", "HOLD")
    confidence = decision.get("confidence", 0)
    
    if signal in ["BUY", "SELL"] and confidence > 75:
        order_side = OrderSide.BUY if signal == "BUY" else OrderSide.SELL
        # Calculate order quantity (e.g., $250 notional)
        notional_usd = 250
        qty = round(notional_usd / latest_price, 6)
        
        order_data = MarketOrderRequest(
            symbol="BTC/USD",
            qty=qty,
            side=order_side,
            time_in_force=TimeInForce.GTC
        )
        
        # Submit order via Alpaca Trading Client
        order = trade_client.submit_order(order_data)
        print(f"Successfully executed {signal} order for {qty} BTC/USD. Order ID: {order.id}")
    else:
        print("No trade executed. Signal was HOLD or confidence was too low.")
except Exception as e:
    print("Error executing Gemini AI strategy:", str(e))
    print("Falling back to local Sovereign OS API execution...")
    try:
        # Fallback: Call the Sovereign OS API directly to run the strategy
        api_url = os.getenv("SOVEREIGN_API_URL", "http://localhost:3000/api/v1/crypto/btc-swing-strategy")
        res = requests.post(api_url, json={"executeOrder": True, "notionalAmount": 250})
        print("Sovereign OS API Response:", res.json())
    except Exception as api_err:
        print("Sovereign OS API Fallback failed:", str(api_err))
````

# Step 5: Visualizing the Swing Trading Strategy

````python
plt.figure(figsize=(14, 7))
plt.plot(df.index, df['close'], label='BTC/USD Close', color='black', alpha=0.6)
plt.plot(df.index, df['ema_short'], label='EMA 12 (Short)', color='blue', alpha=0.8)
plt.plot(df.index, df['ema_long'], label='EMA 26 (Long)', color='orange', alpha=0.8)

# Plot Buy Signals
buy_signals = df[df['buy_signal']]
plt.scatter(buy_signals.index, buy_signals['close'], label='Buy Signal', marker='^', color='green', s=100)

# Plot Sell Signals
sell_signals = df[df['sell_signal']]
plt.scatter(sell_signals.index, sell_signals['close'], label='Sell Signal', marker='v', color='red', s=100)

plt.title('BTC/USD Swing Trading Strategy with EMA Crossover')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend(loc='best')
plt.grid(True)
plt.show()
````
